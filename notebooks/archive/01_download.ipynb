# Early and 2024 samples (superseded)

These notebooks produced the first dataset (earliest 2014 videos) and a 2024 sample.
They are kept for reference. The 2024 sample (`recent_60k.parquet`) will be reused as an
out-of-time check at the end of the project. The main dataset is built in `01c`.

```python
from pathlib import Path
from huggingface_hub import list_repo_files, hf_hub_download
import pyarrow.parquet as pq

ROOT = Path.cwd().parent
RAW = ROOT / "data" / "raw"
REPO = "datasocial/tiktok-5.6B-videos"
TARGET_ROWS = 60_000

files = sorted(
    f for f in list_repo_files(REPO, repo_type="dataset")
    if f.startswith("data/") and f.endswith(".parquet")
)
print(len(files), "files in the repo")
print(files[:3], files[-3:])
```

```python
total = 0
for f in files:
    local = hf_hub_download(REPO, f, repo_type="dataset", local_dir=RAW)
    n = pq.ParquetFile(local).metadata.num_rows
    total += n
    print(f, n, "cumulative:", total)
    if total >= TARGET_ROWS:
        break
```
